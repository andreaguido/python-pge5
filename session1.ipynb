### Variables

### Types
Every value has a type. The type decides what you can do with it.

````python
print(type(20000), type(0.2212), type("female"), type(True))
````

A number stored as text cannot be used in arithmetic. Remove the `#` below and run the cell to see the error, then put the `#` back.

````python
# "20000" + 5
int("20000") + 5     # converting the text to a number fixes it
````

### Lists
Counting starts at **0**. A slice `[a:b]` includes `a` and excludes `b`.

````python
bills = [3913, 3102, 689, 0, 0, 0]   # client 1: Sep, Aug, Jul, Jun, May, Apr
print(bills[0], bills[-1], bills[0:3], len(bills))
````

### Dictionaries
A dictionary maps names (keys) to values. One row of our data is exactly that.

````python
client = {"ID": 1, "LIMIT_BAL": 20000, "SEX": 2, "AGE": 24, "default": 1}
client["AGE"] = 25
print(client["LIMIT_BAL"], client.keys())
````

### Loops and `if`
How many of the last six months was client 1 late on payment?

````python
pay_status = [2, 2, -1, -1, -2, -2]   # months late, Sep -> Apr

late_months = 0
for status in pay_status:
    if status >= 1:
        late_months = late_months + 1

late_months
````

### Functions

````python
def utilisation(bill, limit):
    """Share of the credit limit used by the bill."""
    return bill / limit

print(utilisation(3913, 20000))
print(utilisation(150000, 120000))   # above 1: over the limit?
````

## Parts 2–3 — Importing the data (and reading object syntax)

We load the library once and call it `pd`.

If you get `ImportError: Missing optional dependency 'xlrd'`, the old `.xls` format needs an extra package. Run `!pip install xlrd` in a cell, then restart the kernel (*Kernel → Restart*).

````python
import pandas as pd

path = "data/raw/default of credit card clients.xls"
````

### Failure 2: a silent import problem
The naive import runs without error. Look at the result.

````python
wrong = pd.read_excel(path)
wrong.head(3)
````

````python
wrong.dtypes.head()      # every column is 'object' (text/mixed), not a number
````

### The fix: tell Pandas the header is on row 1 (counting from 0)

````python
df = pd.read_excel(path, header=1)
df.shape
````

### Reading object syntax
- **No parentheses** = attribute, i.e. a fact about the object: `df.shape`, `df.columns`
- **Parentheses** = method, i.e. an action: `df.head()`, `df.describe()`

Read a chain from left to right; each dot acts on the result of the previous step.

````python
df["LIMIT_BAL"].describe().round(0)
````

````python
df.sort_values("LIMIT_BAL", ascending=False).head(3)
````
